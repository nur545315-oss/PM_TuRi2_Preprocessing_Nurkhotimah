# Langkah 1 - Menangani nilai hilang (imputasi median/modus)

```python
import pandas as pd
import numpy as np

data = {
    'usia': [25, 30, np.nan, 35, 28, 40, np.nan, 32, 45, 27],
    'pendapatan': [3000000, 4500000, 5000000, np.nan, 3500000, 7000000, 6000000, np.nan, 8500000, 4000000],
    'pendidikan': ['SMA', 'S1', 'S1', 'S2', 'SMA', 'S2', 'S1', 'S1', 'S2', 'SMA'],
    'kota': ['Kuningan', 'Cirebon', 'Majalengka', 'Kuningan', 'Cirebon', 'Majalengka', 'Kuningan', 'Cirebon', 'Majalengka', 'Kuningan'],
    'membeli': ['Tidak', 'Ya', 'Ya', 'Ya', 'Tidak', 'Ya', 'Ya', 'Tidak', 'Ya', 'Tidak']
}

df = pd.DataFrame(data)

df
df['usia'] = df['usia'].fillna(df['usia'].median())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].median())
```

# Langkah 2 - Memisahkan fitur (X) dan label (Y)

```python
x = df.drop(columns=['membeli'])
y = df['membeli']
```

# Langkah 3 - Encoding kategorikal (ordinal & one-hot)

```python
x['pendidikan'] = x['pendidikan'].map({'SMA': 0, 'S1': 1, 'S2': 2})
x = pd.get_dummies(x, columns=['kota'], dtype=int)
y = y.map({'Tidak':0, 'Ya':1})
```

# Langkah 4 - Membagi dta (SEBELUM scaling)

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.3, random_state=42)
```

# Langkah 5 - Pensklaan (fit pada latih, transform pada latih & uji)

```python
from sklearn.preprocessing import StandardScaler
num = ['usia', 'pendapatan', 'pendidikan']
sc = StandardScaler()
x_train[num] = sc.fit_transform(x_train[num])
x_test[num] = sc.transform(x_test[num])
```

# 1. Mengapa urutan 'split dulu, baru scaling' penting?

Agar informasi dari data uji tidak ikut digunakan saat proses scaling pada data latih, sehingga dapat mencegah data leakage dan menghasilkan evaluasi model yang lebih valid.

# 2. Kapan label encoding dan kapan one-hot encoding?

Label encoding digunakan untuk kategori yang memiliki urutan, sedangkan one-hot encoding digunakan untuk kategori yang tidak memiliki urutan.

# 3. Apa perbedaan normalisasi dan standardisasi?

Normalisasi mengubah nilai data ke rentang tertentu, biasanya 0–1. Sedangkan standardisasi mengubah data berdasarkan mean dan standar deviasi, sehingga data memiliki rata-rata sekitar 0 dan standar deviasi sekitar 1.
